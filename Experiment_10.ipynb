# **Part - A**

```python
print("24BAD048 - Jeevaprakash S")
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

device = (
    "cuda" if torch.cuda.is_available()
    else "mps" if torch.backends.mps.is_available()
    else "cpu"
)

print("Selected device:", device)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)
model.to(device)
model.eval()

messages = [
    {"role": "user", "content": "What is artificial intelligence?"}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
).to(device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=False
    )

response = tokenizer.decode(
    outputs[0][inputs["input_ids"].shape[-1]:],
    skip_special_tokens=True
)
print("Model Response:", response)
```

# **Part - B**

```python
print("24BAD048 - Jeevaprakash S")
def generate_response(user_message, history=None):
    if not user_message or not user_message.strip():
        return "Please enter a message."

    try:
        messages = [
            {
                "role": "system",
                "content": (
                    "You are a helpful AI assistant. "
                    "Give clear, concise, and relevant answers."
                )
            }
        ]

        # Include previous conversation turns
        for turn in (history or [])[-4:]:
            if isinstance(turn, dict):
                if turn.get("role") in ("user", "assistant"):
                    messages.append({
                        "role": turn["role"],
                        "content": turn["content"]
                    })
            elif isinstance(turn, (list, tuple)) and len(turn) == 2:
                if turn[0]:
                    messages.append({
                        "role": "user",
                        "content": turn[0]
                    })
                if turn[1]:
                    messages.append({
                        "role": "assistant",
                        "content": turn[1]
                    })

        messages.append({
            "role": "user",
            "content": user_message.strip()
        })

        inputs = tokenizer.apply_chat_template(
            messages,
            add_generation_prompt=True,
            tokenize=True,
            return_dict=True,
            return_tensors="pt"
        ).to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_new_tokens=150,
                do_sample=False
            )

        answer = tokenizer.decode(
            outputs[0][inputs["input_ids"].shape[-1]:],
            skip_special_tokens=True
        ).strip()

        return answer or "The model did not generate a response."

    except Exception as error:
        print("Runtime error:", error)
        return "Sorry, an error occurred while generating the response."
```

```python
test_prompts = [
    "What is machine learning?",
    "Explain Python lists with an example.",
    "Write a short story about a clever fox.",
    "What is the difference between AI and ML?",
    "Give me three tips to prepare for an exam."
]

for i, prompt in enumerate(test_prompts, start=1):
    print(f"\nTest {i}: {prompt}")
    print("Response:", generate_response(prompt))

print("\nEmpty input test:")
print(generate_response(""))
```
